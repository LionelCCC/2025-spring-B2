# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Separate Subplots :

- Advantages: Allows for a clear, direct comparison of the visitor numbers for each week independently. Any unique patterns or trends within a specific week are easily discernible without being potentially masked by the other week's data. The y-axis scaling can be optimized for each week if the visitor numbers differ significantly.
- Disadvantages: Makes it slightly harder to directly compare the visitor numbers on the same day across the two weeks, as you have to visually scan between the two separate plots. The overall trend or similarity between the two weeks might not be immediately apparent.

Single Plot with Both Series :

- Advantages: Facilitates a direct comparison of visitor numbers for each day across the two weeks. Overlapping or diverging trends between the weeks become immediately obvious. The overall similarity or difference in visitor patterns throughout the week for both weeks is easier to grasp.
- Disadvantages: If the visitor numbers are very close or the lines overlap significantly, it can become difficult to distinguish the data for each week without clear labeling (e.g., different colors and a legend). Unique patterns within a single week might be less prominent if the focus is on the comparison between weeks.

```python
years = np.array([2008, 2009, 2010, 2011, 2012, 2013, 2014])
product_a_prices = np.array([400, 420, 430, 390, 300, 270, 260])
product_b_prices = np.array([360, 400, 410, 380, 290, 260, 250])
product_c_prices = np.array([np.nan, np.nan, 100, 180, 200, 240, 190])
product_d_prices = np.array([np.nan, np.nan, np.nan, 160, 260, 220, 215])
product_e_prices = np.array([np.nan, np.nan, np.nan, np.nan, np.nan, 100, 215])


plt.figure(figsize=(10, 6))
plt.plot(years, product_a_prices, marker='o', linestyle='-', label='Product A')
plt.plot(years, product_b_prices, marker='o', linestyle='-', label='Product B')
plt.plot(years, product_c_prices, marker='o', linestyle='-', label='Product C')
plt.plot(years, product_d_prices, marker='o', linestyle='-', label='Product D')
plt.plot(years, product_e_prices, marker='o', linestyle='-', label='Product E')

plt.xlabel('Year')
plt.ylabel('Average Retail Product Price Per Year ($)')
plt.title('Average Retail Product Price Per Year', fontsize=14)
plt.xticks(years)
plt.yticks(np.arange(0, 550, 100))
plt.ylim(0, 500)

plt.legend(title='Product', loc='lower left', bbox_to_anchor=(0, -0.2), ncol=5)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.grid(axis='y', linestyle='-', alpha=0.7)
plt.tight_layout(rect=[0, 0.05, 1, 1]) # Adjust layout to make room for the legend
plt.show()
```
